```python
import tensorflow as tf
```

```python
a = tf.constant([2,4,6])
b= tf .constant([1,3,5])
```

```python
print("addition of a and b is : ", tf.add(a,b))
print("subtraction of a and b is : ", tf.subtract(a,b))
print("multiplication of a and b is : ", tf.multiply(a,b))
```

```python
print("power of a and b is : ", tf.pow(a,b))
```
