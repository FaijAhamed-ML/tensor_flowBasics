```python
import tensorflow as tf
```

```python
scalar = tf.constant(7)
vector = tf.constant([10, 10])
matrix = tf.constant([[1, 2], [3, 4]])
```

```python
print("Scalar:", scalar)
print("Vector:", vector)
print("Matrix:", matrix)
```

```python
original_tensor = tf.constant([[1, 2], [3, 4]])
reshaped_tensor = tf.reshape(original_tensor, [4, 1])
```

```python
print ("Original Tensor:\n", original_tensor)
print ("Reshaped Tensor:\n", reshaped_tensor)
```

```python
a = tf.constant([1,2,3])
b= tf.constant([4,5,6])
sum_result = tf.add(a, b)
print("Sum of tensors a and b:", sum_result)
```

```python
tensor = tf.constant([[1, 2, 3], [4, 5, 6]])
```

```python
element = tensor[0, 1]
print("Element at position (0, 1):", element)
```

```python
row = tensor[1]
colomn = tensor[:, 2]
print("Row at index 1:", row)
print("Column at index 2:", colomn)
```
