```python
import tensorflow as tf
```

```python
Scalar = tf.constant(7)
vector = tf.constant([10, 10.8, 8.0])
matrix = tf.constant([[1, 2], [3, 4]])
```

```python
print("Scalar:", Scalar)
print("Vector:", vector)
print("Matrix:", matrix)
```

```python
var = tf.Variable(10)
print("Variable:", var)
```

```python
var.assign(20)
print("Updated Variable:", var)
```

```python
a = tf.constant([[1, 2], [3, 4]])
b = tf.constant([[5, 6], [7, 8]])
c = tf.add(a, b)
print("Addition Result:\n", c)
```

```python
a = tf.variable([[1, 2], [3, 4]])
b = tf.variable([[5, 6], [7, 8]])
c = tf.multiply(a, b)
print("Multiplication Result:\n", c)
```

```python
a = tf.constant([[1, 2], [3, 4]])
b = tf.constant([[5, 6], [7, 8]])
c = tf.multiply(a, b)
print("Multiplication Result:\n", c)
```
