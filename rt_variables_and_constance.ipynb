```python
import tensorflow as tf
```

```python
a = tf.constant(5)
b = tf.constant([1.0,2.0,3.0])
```

```python
print("scalar constant a:", a)
print("vector constant b:", b)
```

```python
x = tf.Variable(10)
y = tf.Variable([1.0,2.0,3.0])
```

```python
print ("scalar variable x:", x)
print ("vector variable y:", y)
```

```python
counter = tf.Variable(0)
```

```python
counter.assign_add(1)
```

```python
print("counter after increment:", counter.numpy())
```

```python
counter.assign(10)
```

```python
print("counter after reassigning to 10:", counter.numpy())
```
