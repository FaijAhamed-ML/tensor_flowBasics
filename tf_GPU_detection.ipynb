```python
import tensorflow as tf
```

```python
gpu_devices = tf.config.list_physical_devices('GPU')
if len(gpu_devices) == 0:
    print("No GPU devices found.")
else:
    print("GPU devices found:")
    for device in gpu_devices:
        print(device)
```
